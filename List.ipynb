```python
names=["Priya", "Prajyoti", "Bhargavi","Sneha","Gayatri"]
print(names)
```

```python
print(len(names)) #checks length of the elements presesnt in list
```

```python
print("Gayatri" in names) # specific name present in list
```

```python
print("Rahul" not in names) # specific name not present  in list
```

```python
names.sort() #sorts in asc order
print(names)
```

```python
names.sort(reverse=True) #sorts in desc order
print(names)
```

```python
print(names)
```

```python
new_names=sorted(names) #gave new list in reverse order
print(new_names)
print(names)
```
